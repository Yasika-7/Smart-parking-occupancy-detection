```python
import json

ann_path = os.path.join(train_dir, "_annotations.coco.json")

with open(ann_path) as f:
    coco = json.load(f)

print("Categories:", coco['categories'])
print("Total images:", len(coco['images']))
print("Total annotations (bounding boxes):", len(coco['annotations']))
print("\nSample annotation:", coco['annotations'][5])
```

```python
from PIL import Image

os.makedirs('/content/classified/free', exist_ok=True)
os.makedirs('/content/classified/occupied', exist_ok=True)

cat_map = {c['id']: c['name'] for c in coco['categories']}
img_map = {img['id']: img['file_name'] for img in coco['images']}

count_free, count_occupied, skipped = 0, 0, 0

for ann in coco['annotations']:
    cat_id = ann['category_id']
    cat_name = cat_map.get(cat_id, "")

    # Skip the supercategory "cars-park" (id 0) - only care about Occupied/Vacant
    if cat_name not in ["Occupied", "Vacant"]:
        skipped += 1
        continue

    img_file = img_map[ann['image_id']]
    img_path = os.path.join(train_dir, img_file)

    if not os.path.exists(img_path):
        skipped += 1
        continue

    x, y, w, h = ann['bbox']
    x, y, w, h = int(x), int(y), int(w), int(h)

    if w <= 5 or h <= 5:   # skip tiny/invalid boxes
        skipped += 1
        continue

    try:
        img = Image.open(img_path).convert("RGB")
        cropped = img.crop((x, y, x + w, y + h))

        if cat_name == "Vacant":
            cropped.save(f'/content/classified/free/{count_free}.jpg')
            count_free += 1
        elif cat_name == "Occupied":
            cropped.save(f'/content/classified/occupied/{count_occupied}.jpg')
            count_occupied += 1
    except Exception:
        skipped += 1

print(f"Free (Vacant): {count_free}")
print(f"Occupied: {count_occupied}")
print(f"Skipped: {skipped}")
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

IMG_SIZE = (64, 64)
BATCH_SIZE = 32
data_dir = "/content/classified"

train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

class_names = train_ds.class_names
print("Classes:", class_names)   # ['free', 'occupied'] varanum
```

```python
normalization_layer = layers.Rescaling(1./255)
train_ds = train_ds.map(lambda x, y: (normalization_layer(x), y))
val_ds = val_ds.map(lambda x, y: (normalization_layer(x), y))

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.cache().prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.cache().prefetch(buffer_size=AUTOTUNE)
```

```python
model = models.Sequential([
    layers.Input(shape=(64, 64, 3)),
    layers.Conv2D(16, 3, activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(32, 3, activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10
)
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(history.history['accuracy'], label='Train Accuracy')
axes[0].plot(history.history['val_accuracy'], label='Val Accuracy')
axes[0].set_title('Model Accuracy')
axes[0].set_xlabel('Epoch')
axes[0].legend()

axes[1].plot(history.history['loss'], label='Train Loss')
axes[1].plot(history.history['val_loss'], label='Val Loss')
axes[1].set_title('Model Loss')
axes[1].set_xlabel('Epoch')
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
def representative_dataset():
    for images, _ in train_ds.take(50):
        for img in images:
            yield [tf.expand_dims(img, axis=0)]

converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.representative_dataset = representative_dataset
converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
converter.inference_input_type = tf.uint8
converter.inference_output_type = tf.uint8

tflite_quant = converter.convert()

with open('/content/parking_model_int8.tflite', 'wb') as f:
    f.write(tflite_quant)

print(f"Quantized size: {os.path.getsize('/content/parking_model_int8.tflite')/1024:.1f} KB")
```

```python
import numpy as np

interpreter = tf.lite.Interpreter(model_path='/content/parking_model_int8.tflite')
interpreter.allocate_tensors()
inp = interpreter.get_input_details()[0]
out = interpreter.get_output_details()[0]

correct, total = 0, 0
for images, labels in val_ds.take(20):   # 20 batches = 640 images
    for img, label in zip(images.numpy(), labels.numpy()):
        interpreter.set_tensor(inp['index'], np.expand_dims(img.astype(np.uint8), 0))
        interpreter.invoke()
        pred = interpreter.get_tensor(out['index'])[0][0]
        pred_class = 1 if pred > 128 else 0    # uint8 output: 0-255
        correct += int(pred_class == label)
        total += 1

print(f"Quantized model accuracy: {correct/total*100:.2f}% ({total} images)")
```

```python
!pip install roboflow -q

import os, json, random, shutil, numpy as np, tensorflow as tf
from collections import defaultdict
from PIL import Image
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

tf.keras.utils.set_random_seed(42)
IMG_SIZE = (64, 64)
OUT = '/content/results'
MODEL = '/content/parking_cnn_model.keras'
os.makedirs(OUT, exist_ok=True)
log = []

def say(*a):
    t = " ".join(str(x) for x in a)
    print(t)
    log.append(t)

train_dir = '/content/CNRPark+EXT-4/train'
ann_path = os.path.join(train_dir, '_annotations.coco.json')
if not os.path.exists(ann_path):
    from roboflow import Roboflow
    rf = Roboflow(api_key="PASTE_YOUR_KEY_HERE")
    ws = rf.workspace("autonomousparking")
    ws.project("cnrpark-ext-cnziv-f7ygy").version(4).download("coco")

coco = json.load(open(ann_path))
cat_map = {c['id']: c['name'] for c in coco['categories']}
img_map = {i['id']: i['file_name'] for i in coco['images']}
by_img = defaultdict(list)
for a in coco['annotations']:
    by_img[a['image_id']].append(a)

if not os.path.exists('/content/classified/occupied'):
    os.makedirs('/content/classified/free', exist_ok=True)
    os.makedirs('/content/classified/occupied', exist_ok=True)
    cf = 0
    co = 0
    for iid, anns in by_img.items():
        path = os.path.join(train_dir, img_map[iid])
        if not os.path.exists(path):
            continue
        im = Image.open(path).convert("RGB")
        for a in anns:
            name = cat_map.get(a['category_id'], "")
            if name not in ("Occupied", "Vacant"):
                continue
            x, y, w, h = [int(v) for v in a['bbox']]
            if w <= 5 or h <= 5:
                continue
            crop = im.crop((x, y, x + w, y + h))
            if name == "Vacant":
                crop.save('/content/classified/free/%d.jpg' % cf)
                cf += 1
            else:
                crop.save('/content/classified/occupied/%d.jpg' % co)
                co += 1
    say("Cropped: free", cf, "occupied", co)

train_ds = tf.keras.utils.image_dataset_from_directory(
    '/content/classified', validation_split=0.2, subset="training",
    seed=123, image_size=IMG_SIZE, batch_size=32)
val_ds = tf.keras.utils.image_dataset_from_directory(
    '/content/classified', validation_split=0.2, subset="validation",
    seed=123, image_size=IMG_SIZE, batch_size=32)
say("Classes:", train_ds.class_names)
train_ds = train_ds.cache().prefetch(tf.data.AUTOTUNE)
val_ds = val_ds.cache().prefetch(tf.data.AUTOTUNE)

if os.path.exists(MODEL):
    model = tf.keras.models.load_model(MODEL)
    say("Saved model loaded")
else:
    model = models.Sequential([
        layers.Input(shape=(64, 64, 3)),
        layers.Rescaling(1./255),
        layers.Conv2D(16, 3, activation='relu'),
        layers.MaxPooling2D(),
        layers.Conv2D(32, 3, activation='relu'),
        layers.MaxPooling2D(),
        layers.Conv2D(64, 3, activation='relu'),
        layers.MaxPooling2D(),
        layers.Flatten(),
        layers.Dense(64, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(1, activation='sigmoid'),
    ])
    model.compile(optimizer='adam', loss='binary_crossentropy',
                  metrics=['accuracy'])
    hist = model.fit(train_ds, validation_data=val_ds, epochs=6)
    h = hist.history
    for i in range(6):
        say("Epoch", i + 1,
            "train_acc %.4f" % h['accuracy'][i],
            "train_loss %.4f" % h['loss'][i],
            "val_acc %.4f" % h['val_accuracy'][i],
            "val_loss %.4f" % h['val_loss'][i])
    model.save(MODEL)
    fig, ax = plt.subplots(1, 2, figsize=(9, 3.3))
    ax[0].plot(h['loss'], 'k-', label='Training')
    ax[0].plot(h['val_loss'], 'k--', label='Validation')
    ax[0].set_title('Loss')
    ax[0].legend()
    ax[1].plot(h['accuracy'], 'k-', label='Training')
    ax[1].plot(h['val_accuracy'], 'k--', label='Validation')
    ax[1].set_title('Accuracy')
    ax[1].legend()
    plt.tight_layout()
    plt.savefig(OUT + '/training_curves.png', dpi=200)
    plt.show()

print("CELL A DONE")
```

```python
def rep_data():
    for images, _ in train_ds.take(50):
        for img in images:
            yield [tf.expand_dims(img, axis=0)]

conv = tf.lite.TFLiteConverter.from_keras_model(model)
open('/content/parking_model.tflite', 'wb').write(conv.convert())

conv = tf.lite.TFLiteConverter.from_keras_model(model)
conv.optimizations = [tf.lite.Optimize.DEFAULT]
conv.representative_dataset = rep_data
conv.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
conv.inference_input_type = tf.uint8
conv.inference_output_type = tf.uint8
open('/content/parking_model_int8.tflite', 'wb').write(conv.convert())

say("Keras file KB:", round(os.path.getsize(MODEL) / 1024, 1))
say("TFLite float KB:", round(os.path.getsize('/content/parking_model.tflite') / 1024, 1))
say("TFLite int8 KB:", round(os.path.getsize('/content/parking_model_int8.tflite') / 1024, 1))

interp = tf.lite.Interpreter(model_path='/content/parking_model_int8.tflite')
interp.allocate_tensors()
inp = interp.get_input_details()[0]
out = interp.get_output_details()[0]

q_true = []
q_pred = []
for images, labels in val_ds:
    for img, lab in zip(images.numpy(), labels.numpy()):
        interp.set_tensor(inp['index'], np.expand_dims(img.astype(np.uint8), 0))
        interp.invoke()
        q_pred.append(int(interp.get_tensor(out['index'])[0][0] > 128))
        q_true.append(int(lab))
q_true = np.array(q_true)
q_pred = np.array(q_pred)
say("[INT8] accuracy on ALL %d validation images: %.2f%%" % (len(q_true), (q_true == q_pred).mean() * 100))
say(classification_report(q_true, q_pred, target_names=['free', 'occupied'], digits=4))
say("INT8 confusion matrix:")
say(confusion_matrix(q_true, q_pred))

data = open('/content/parking_model_int8.tflite', 'rb').read()
rows = []
for i in range(0, len(data), 12):
    rows.append("  " + ", ".join("0x%02x" % b for b in data[i:i + 12]))
header = "#ifndef PARKING_MODEL_H\n#define PARKING_MODEL_H\n\n"
header += "alignas(8) const unsigned char parking_model[] = {\n"
header += ",\n".join(rows)
header += "\n};\n\nconst unsigned int parking_model_len = %d;\n\n#endif\n" % len(data)
open('/content/parking_model.h', 'w').write(header)

print("CELL C DONE")
```
